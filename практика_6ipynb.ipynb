```python
def bayes(prior, p_obstacle, p_clear):
    a = prior * p_obstacle
    b = (1 - prior) * p_clear
    return a / (a + b)
```

```python
print("D1 нәтижелері:")

for prior in [0.10, 0.30, 0.50, 0.70]:
    result = bayes(prior, 0.80, 0.20)
    print(prior, "->", round(result, 4))
```

```python
print("\nD2 нәтижелері:")

old = bayes(0.30, 0.80, 0.20)
new = bayes(0.30, 0.80, 0.40)

print("Бастапқы:", round(old, 4))
print("Жаңа:", round(new, 4))
```

```python
print("\nD3 нәтижелері:")

observations = ["BLOCKED", "CLEAR", "BLOCKED", "BLOCKED"]

emission = {
    "BLOCKED": (0.80, 0.20),
    "CLEAR": (0.20, 0.80)
}

p = 0.30

for t, reading in enumerate(observations, start=1):
    if t == 1:
        predicted = p
    else:
        predicted = 0.70 * p + 0.20 * (1 - p)

    a, b = emission[reading]
    p = bayes(predicted, a, b)

    print(
        "t" + str(t),
        reading,
        "Болжам:", round(predicted, 4),
        "P(O):", round(p, 4),
        "P(C):", round(1 - p, 4)
    )
```
